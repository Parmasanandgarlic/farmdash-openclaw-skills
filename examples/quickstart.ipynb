# FarmDash for Agents — 5-minute trial (no key, no wallet)

FarmDash is DeFi intelligence for AI agents: protocol discovery (Trail Heat),
risk scoring (FarmScore), swap/perps planning with **explicitly user-signed**
payloads. Zero-custody: agents plan, humans sign.

This notebook hits only the **free, keyless Scout sandbox** — safe to run anywhere.
Docs: https://www.farmdash.one/agents · Skills: https://github.com/Parmasanandgarlic/farmdash-openclaw-skills

```python
import urllib.request, json

def get(path):
    req = urllib.request.Request(
        'https://www.farmdash.one' + path,
        headers={'User-Agent': 'FarmDash-demo-notebook/1.0'})
    with urllib.request.urlopen(req, timeout=30) as r:
        return r.status, r.read()

status, body = get('/api/status')
print('GET /api/status ->', status, f'({len(body)} bytes)')
print(json.dumps(json.loads(body), indent=1)[:600])
```

## 1. Discover protocols (mock sandbox)

Same shape as production; values are deterministic fixtures (`mock mode is for development only`).

```python
status, body = get('/api/v1/agent/protocols?mock=true')
d = json.loads(body)
print('tier:', d['tier'], '| mode:', d['mode'], '| count:', d['count'])
for p in d['data']:
    print(f"- {p['protocol_name']} ({'/'.join(p['chains'])}) "
          f"score={p['discovery_heuristic_score']} sybil={p['sybil_risk']} tvl={p['tvl']}")
```

## 2. Machine-readable surfaces

Agents don't scrape HTML — they read `mcp.json` and the OpenAPI doc.

```python
status, body = get('/.well-known/mcp.json')
mcp = json.loads(body)
print('mcp.json ->', status, f'({len(body)//1024} KB)')
print('description:', mcp.get('description', '')[:120])
print('top-level keys:', sorted(mcp.keys()))
print('tools:', len(mcp.get('tools', [])))

status, body = get('/agents/openapi.yaml')
text = body.decode()
ops = text.count('operationId:')
print(f'\nopenapi.yaml -> {status} ({len(body)//1024} KB, ~{ops} operations)')
```
